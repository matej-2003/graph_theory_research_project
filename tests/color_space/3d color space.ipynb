```python
import numpy as np
import plotly.graph_objects as go

def plot_rgb_space(samples=3000):
    # 1. Generate random RGB colors
    # We use random sampling to see the "volume" without creating a heavy grid
    r = np.random.randint(0, 256, samples)
    g = np.random.randint(0, 256, samples)
    b = np.random.randint(0, 256, samples)

    # 2. Create the color strings for Plotly
    # Format: 'rgb(r, g, b)'
    colors = [f'rgb({r[i]},{g[i]},{b[i]})' for i in range(samples)]

    # 3. Create the 3D Scatter plot
    fig = go.Figure(data=[go.Scatter3d(
        x=r,
        y=g,
        z=b,
        mode='markers',
        marker=dict(
            size=5,
            color=colors,  # Color the point by its own RGB value
            opacity=0.8
        )
    )])

    # 4. Layout formatting
    fig.update_layout(
        title="RGB Color Space (Cartesian)",
        scene=dict(
            xaxis=dict(title='Red', range=[0, 255]),
            yaxis=dict(title='Green', range=[0, 255]),
            zaxis=dict(title='Blue', range=[0, 255]),
            aspectmode='cube'
        ),
        margin=dict(l=0, r=0, b=0, t=50)
    )

    fig.show()

plot_rgb_space()
```
