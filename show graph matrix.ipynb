```python
def display_graphs(graphs, figsize=(15, 15), save_path=None):
    def axes_division(x):
        factor1 = 1
        for i in range(1, int(math.sqrt(x)) + 1):
            if x % i == 0:
                factor1 = i
        return int(factor1), int(x / factor1)

    rows, cols = axes_division(len(graphs))
    fig, axes = plt.subplots(rows, cols, figsize=figsize)
    axes = np.array(axes).flatten()

    for (graph_tuple, ax) in zip(graphs, axes):
        G, n, a = graph_tuple
        if a == 2:
            display_graph2(G, show_title=True, pos=prism_pos(G), ax=ax)
        else:
            display_graph2(G, show_title=True, pos=mobius_pos(G), ax=ax)

        ax.set_title(f"Graph C({n}, {a})")
        ax.set_aspect("equal")

    plt.tight_layout()

    if save_path is not None:
        fig.savefig(save_path, dpi=300, bbox_inches="tight")

    plt.show()

search_space = 24
graphs = []

for i in range(4, search_space+1, 2):
    n_graph_list = generate_graphs(i)
    graphs += n_graph_list

print(len(graphs))

# display_graphs(graphs)
display_graphs(graphs, save_path="all_graphs.png")

# save_graphs(graphs)
```
