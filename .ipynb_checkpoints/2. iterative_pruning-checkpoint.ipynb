```python
%load_ext autoreload
%autoreload 2

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import math

# import import_ipynb
from graph_functions import *
from iterative_pruning_func import *
```

```python
N = 60
graphs = []

for i in range(4, N, 2):
    q = generate_graphs(i)
    graphs.append(q)
```

```python
G, n, _ = generate_graphs(26)[1]     # prism
fully_infect(G)

def test_algorithem(G_):
    G = G_.copy()

    # start with a fully infected graph
    fully_infect(G)
    
    # 1. stage
    minimise_infection(G, 0)
    pc = calc_graph_values(G)
    
    for iteration in range(100):
        if len(pc) == 0:
            break
        
        e = pc[0]
        G.nodes[pc[0]]["infected"] = False
        pc = calc_graph_values(G)
        
        H, infected_nodes, healthy_nodes = fully_endemic(G)
        if healthy_nodes:
            print("Graph healhy nodes ramaining ", healthy_nodes)
            print("FAIL")

    # print("iterations: ", iteration)

    min_infection_number = 0
    for x in G.nodes():
        if G.nodes[x].get("infected"):
            min_infection_number += 1

    # print("Minimal infection number: ", min_infection_number)
    return min_infection_number, G


print()
print()
FIGSIZE=(6, 3)

for u in graphs:
    for G, n, a in u:
        print(f"C(n={n}, a={a}) ", end="")
        expected_value = math.ceil(n/4)
        calc_min_IN, H = test_algorithem(G)
        
        if a == 1:
            print("MOBIUS ")
            display_graph_stage(H, prism=False, figsize=FIGSIZE)
            print("MOBIUS EXPECTED Minimal infection number: " , expected_value)
        else:
            print("PRISM ")
            display_graph_stage(H, prism=True, figsize=FIGSIZE)
            print("MOBIUS EXPECTED Minimal infection number: ", expected_value + 1)

        print("CALCULATED Minimal infection number: " + str(calc_min_IN))
        # print("Result status: ", expected_value == calc_min_IN)
        print()
```
