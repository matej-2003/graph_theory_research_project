```python
def has_common_neighbor(G, u, v):
    return len(set(G.neighbors(u)) & set(G.neighbors(v))) > 0

def prune_by_common_neighbors_stages(G_):
    G = G_.copy()
    
    stages = []
    
    infected_nodes = set(G.nodes)
    infected_nodes.remove(0)
    healthy_nodes = {0}

    # stage 0: initial state
    stages.append(G.copy())
    to_visit = {0}

    while True:
        if len(to_visit) == 0:
            break
    
        e = list(to_visit)[0]
        to_visit.remove(e)
        print(e)
        
        N_e = G.neighbors(e)
        for x in N_e:
            N_x = G.neighbors(x)
            for y in N_x:
                if y not in N_e:
                    to_visit.add(y)

        print(to_visit)
    return stages


def display_pruning_stages(stages):
    graphs = []
    for i, H in enumerate(stages):
        graphs.append((H, len(H.nodes), i))

    display_graphs(
        graphs,
        figsize=(3 * len(graphs), 3),
        show_titles=True,
        force_row=True,
        with_labels=True,
    )
# pos=lambda G: prism_pos(G)
```

```python
x = generate_graphs(10)
G, _, _ = x[1]

# G is fully infected
seed_nodes = [0]

stages = prune_by_common_neighbors_stages(G)
print(stages)
display_pruning_stages(stages)

```
