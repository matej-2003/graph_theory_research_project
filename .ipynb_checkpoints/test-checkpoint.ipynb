```python
def print_bin_list(k, n, bin_list):
    if len(bin_list) == n:
        return

    a = bin_list[:] + [0]
    
    if sum(a) == k:
        yield a + [0] * (n-len(bin_list))
    if sum(a) + n - len(bin_list) < k:
        return
    b = bin_list[:] + [1]
    yield from print_bin_list(k, n, a)

    if sum(b) == k:
        yield b + [0] * (n-len(bin_list))

    if sum(b) + n - len(bin_list):
        return

    yield from print_bin_list(k, n, b)

g = print_bin_list(2, 4, [])


for i in g:
    print(i)
```
