```python
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import math

import import_ipynb
from graph_functions import *
```

```python
x = generate_graphs(10)
G, _, _ = x[1]
```

```python
def color_distance_partitions(G, v = 0, figsize=(3, 3), with_labels=False, pos=None):
    # get BFS layers
    layers = list(nx.bfs_layers(G, v))
    
    # assign a color index to each node based on distance
    node_to_layer = {}
    for i, layer in enumerate(layers):
        for node in layer:
            node_to_layer[node] = i
    
    
    print(layers)
    # layout (spring works fine, radial is optional)
    if pos == None:
        pos_ = nx.spring_layout(G, seed=100)
    else:
        pos_ = pos(G)
    
    # colors by layer
    node_colors = [node_to_layer[n] for n in G.nodes()]
    
    plt.figure(figsize=figsize)
    edge_colors = [
        'red' if G.nodes[n].get("infected", False) else 'black'
        for n in G.nodes()
    ]
    nx.draw(
        G,
        pos_,
        with_labels=with_labels,
        node_color=node_colors,
        edgecolors=edge_colors,
        linewidths=2,
        cmap=plt.cm.viridis,
        node_size=150,
        font_size=10,
    )
    
    plt.title(f"Distance partitions from v = {v}")
    plt.show()


def visualize_distance_partitions(G, v = 0, figsize=(8, 4), with_labels=False):
    # compute distance layers
    layers = list(nx.bfs_layers(G, v))
    
    # build custom positions
    pos = {}
    x_spacing = 2.0
    y_spacing = 1.0
    
    for i, layer in enumerate(layers):
        x = i * x_spacing
        y_start = -(len(layer) - 1) * y_spacing / 2
    
        for j, node in enumerate(sorted(layer)):
            y = y_start + j * y_spacing
            pos[node] = (x, y)
    
    # color by layer
    node_colors = []
    for node in G.nodes():
        for i, layer in enumerate(layers):
            if node in layer:
                node_colors.append(i)
                break
    edge_colors = [
        'red' if G.nodes[n].get("infected", False) else 'black'
        for n in G.nodes()
    ]
    
    plt.figure(figsize=figsize)
    nx.draw(
        G,
        pos,
        with_labels=with_labels,
        node_color=node_colors,
        edgecolors=edge_colors,
        linewidths=2,
        cmap=plt.cm.viridis,
        node_size=150,
        font_size=10,
    )
    
    plt.title(f"Distance-layer visualization from v = {v}")
    plt.axis("off")
    plt.show()



for i in range(len(G.nodes)):
    G.nodes[i]["infected"] = 1

G.nodes[0]["infected"] = False

color_distance_partitions(G, pos=prism_pos)
visualize_distance_partitions(G, 0, figsize=(4, 2))
# for i in G.nodes:
#     visualize_distance_partitions(G, i, figsize=(4, 2))
```
