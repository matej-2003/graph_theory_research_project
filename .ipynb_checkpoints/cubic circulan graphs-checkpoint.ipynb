```python
G1 = nx.Graph()

def set_Z(start=0):
    i = start
    while True:
        yield i
        i += 1

def cyclic_Z(start=0, n=10):
    gen = set_Z(start)
    while True:
        i = next(gen) % n
        yield i


S = cyclic_Z(0, 3)
print(next(S))
print(next(S))
print(next(S))
print(next(S))

```

```python
from pyvis.network import Network
import networkx as nx

a = 7
n = 16
# n = 6

print(f"a={a} n={n} n/2={n/2}")
print("1 <= a < n/2 : " , 1 <= a < n/2)


Z = list(range(0, n))
S = [a, -a, n/2]
E = []
print(S)


for v in Z:
    for s in S:
        u = int(v + s) % n
        E.append((v, u))

print(E)


G = nx.Graph()

# G.add_nodes_from(Z)
G.add_edges_from(E)

for node in G.nodes():
    G.nodes[node]['label'] = str(node)


net = Network(
    notebook=True,
    cdn_resources='in_line',   # <-- important
    height="600px",
    width="100%"
)

net.from_nx(G)
# net.barnes_hut()
# net.force_atlas_2based()
# pos = nx.spring_layout(G, seed=42)

net.show("graph_example.html")
# nx.draw(G, with_labels = True)

```

```python
import numpy as np

n = 16
n_half = int(n/2)

for a in range(n_half):
    gcd = np.gcd(a, n_half)
    condition = (1 <= a < n_half and gcd == 1)
    
    if condition:
        print(f"1 <= {a} < {n/2} && {gcd} == 1 : gcd={gcd} a={a} n={n}" , condition)
```

```python
n = 16
a = 7

G = nx.circulant_graph(16, [a, -a, int(n/2)])


for node in G.nodes():
    G.nodes[node]['label'] = str(node)
net = Network(
    notebook=True,
    cdn_resources='in_line',   # <-- important
    height="600px",
    width="100%"
)

net.from_nx(G)

# net.barnes_hut()
# net.force_atlas_2based()
# pos = nx.spring_layout(G, seed=42)


# print(G.nodes)


# for node in G.nodes:
#     print(f"N(f{node}) = ", list(nx.all_neighbors(G, node)))


# n1 = G.nodes[0]
# n1["infected"] = True
# print(n1.get("infected"))
print(G)


G.nodes[0]["infected"]= True
G.nodes[1]["infected"]= True

import time

def is_infectable(node):
    node_N = list(nx.all_neighbors(G, node))
    if len(node_N) < 2:
        return False
    infected_N_counter = 0
    for i_neighbor in node_N:
        if G.nodes[i_neighbor].get("infected"):
            infected_N_counter += 1
    return (infected_N_counter >= 2)



def fully_endemic(G):
    infected_nodes = []
    healthy_nodes = [n for n in G.nodes]

    itter_number = 0
    while len(healthy_nodes) > 0:
        for node in healthy_nodes:
            if is_infectable(node):
                G.nodes[node]["infected"] = True
                infected_nodes.append(node)
                healthy_nodes.remove(node)

        itter_number+=1
        
        print("itter number=", itter_number, "infected_nodes:", infected_nodes, "healthy_nodes", healthy_nodes)
        time.sleep(1)
        break



fully_endemic(G)

net.show("graph_example.html")
```
