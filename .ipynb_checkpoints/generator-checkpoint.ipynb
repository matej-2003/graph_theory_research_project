```python
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import math

def generate_graphs(n):
	n_half = int(n/2)
	graphs = []
	
	def add_candidate_graph(H, a):
		nonlocal graphs
		for (G_i, _, _) in graphs:
			if nx.is_isomorphic(G_i, H):
				return
		graphs.append((H, n, a))
	
	for a in range(n_half):
		gcd = np.gcd(a, n_half)
		condition = (1 <= a < n_half and gcd == 1)
		
		if condition:
			g = nx.circulant_graph(n, [a, -a, n_half])
			add_candidate_graph(g, a)
			# print(f"a={a} n={n}")
	
	print(f"number of graphs for n={n} a={a}:", len(graphs))
	return graphs
```

```python
def partitions(n, k, p=1):
    if n < p: return
    elif k == 0:
        if n == 0: yield []
    elif k == 1:
        if n != 0: yield [n]
    else:
        for r in partitions(n-p, k-1, p):
            yield [p] + r
        yield from partitions(n, k, p + 1)


def infect_cycle(n, k):
    for l in range(1, k + 1):
        for r in partitions(k, l):
            for p in partitions(n - k, l):
                s = []
                for b, b2 in zip(r, p):
                    # s += "1" * int(b) + "0" * int(b2)
                    s.extend([1] * int(b))
                    s.extend([0] * int(b2))
                yield s
```
